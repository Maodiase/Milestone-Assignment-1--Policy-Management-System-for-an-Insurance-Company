```python
from datetime import datetime
from enum import Enum

class ProductType(Enum):
    LIFE_INSURANCE = "life_insurance"
    HEALTH_INSURANCE = "health_insurance"
    CAR_INSURANCE = "car_insurance"
    HOME_INSURANCE = "home_insurance"

class Product:
    _id_counter = 5000
    
    def __init__(self, product_name, product_type, premium, coverage_amount, description=""):
        self.product_id = self._generate_id()
        self.product_name = product_name
        self.product_type = product_type
        self.premium = premium
        self.coverage_amount = coverage_amount
        self.description = description
        self.created_date = datetime.now()
        self.active = True
    
    def _generate_id(cls):
        cls._id_counter += 1
        return f"PRD{cls._id_counter}"
    
    def update_premium(self, new_premium):
        if new_premium <= 0:
            print("Premium must be greater than zero.")
            return False
        
        old_premium = self.premium
        self.premium = new_premium
        print(f"Premium updated from ${old_premium:.2f} to ${new_premium:.2f} for {self.product_name}.")
        return True
    
    def update_coverage(self, new_coverage):
        if new_coverage <= 0:
            print("Coverage amount must be greater than zero.")
            return False
        
        old_coverage = self.coverage_amount
        self.coverage_amount = new_coverage
        print(f"Coverage updated from ${old_coverage:.2f} to ${new_coverage:.2f} for {self.product_name}.")
        return True
    
    def suspend(self):
        if not self.active:
            print(f"Product {self.product_name} is already suspended.")
            return
        self.active = False
        print(f"Product '{self.product_name}' has been suspended.")
    
    def reactivate(self):
        if self.active:
            print(f"Product {self.product_name} is already active.")
            return
        
        self.active = True
        print(f"Product '{self.product_name}' has been reactivated.")
    
    def get_product_details(self):
        return {
            "product_id": self.product_id,
            "product_name": self.product_name,
            "product_type": self.product_type.value,
            "premium": self.premium,
            "coverage_amount": self.coverage_amount,
            "description": self.description,
            "created_date": self.created_date.strftime("%Y-%m-%d %H:%M:%S"),
            "status": "Active" if self.active else "Suspended"
        }
    
    def __repr__(self):
        status = "Active" if self.active else "Suspended"
        return f"Product(ID={self.product_id}, Name={self.product_name}, Premium=${self.premium:.2f}, Status={status})"
```
