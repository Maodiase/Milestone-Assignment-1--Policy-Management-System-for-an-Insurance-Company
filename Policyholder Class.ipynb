```python
from datetime import datetime
from enum import Enum

class PolicyStatus(Enum):
    ACTIVE = "active"
    SUSPENDED = "suspended"
    INACTIVE = "inactive"

class Policyholder:
    _id_counter = 1000
    
    def __init__(self, name, email, phone, address):
        self.policyholder_id = self._generate_id()
        self.name = name
        self.email = email
        self.phone = phone
        self.address = address
        self.status = PolicyStatus.ACTIVE
        self.registration_date = datetime.now()
        self.products = {}  
        self.payment_history = []
        
    def _generate_id(cls):
        cls._id_counter += 1
        return f"PH{cls._id_counter}"
    
    def register_product(self, product):
        self.products[product.product_id] = product
        print(f"Product '{product.product_name}' registered for {self.name}.")
        return True
    
    def suspend(self, reason=""):
        if self.status == PolicyStatus.SUSPENDED:
            print(f"Policyholder {self.name} is already suspended.")
            return
        
        self.status = PolicyStatus.SUSPENDED
        print(f"Policyholder {self.name} has been suspended. Reason: {reason if reason else 'No reason provided'}")
    
    def reactivate(self):
        if self.status == PolicyStatus.ACTIVE:
            print(f"Policyholder {self.name} is already active.")
            return   
        self.status = PolicyStatus.ACTIVE
        print(f"Policyholder {self.name} has been reactivated.")
    
    def add_payment_record(self, payment_record):
        self.payment_history.append(payment_record)
    
    def get_account_details(self):
        products_list = [
            {
                "product_id": pid,
                "product_name": p.product_name,
                "premium": p.premium
            }
            for pid, p in self.products.items()
        ]
        
        return {
            "policyholder_id": self.policyholder_id,
            "name": self.name,
            "email": self.email,
            "phone": self.phone,
            "address": self.address,
            "status": self.status.value,
            "registration_date": self.registration_date.strftime("%Y-%m-%d %H:%M:%S"),
            "products": products_list,
            "total_premium": sum(p.premium for p in self.products.values()),
            "payment_count": len(self.payment_history)
        }
    
    def __repr__(self):
        return f"Policyholder(ID={self.policyholder_id}, Name={self.name}, Status={self.status.value})"
```
