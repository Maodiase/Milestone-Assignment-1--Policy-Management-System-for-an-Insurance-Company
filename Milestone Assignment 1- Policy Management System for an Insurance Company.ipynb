```python
def display_header(title):

    print("\n" + "="*70)
    print(f" {title}")
    print("="*70)

def display_account_details(policyholder):
    
    details = policyholder.get_account_details()
    print(f"\nPolicyholder ID: {details['policyholder_id']}")
    print(f"Name: {details['name']}")
    print(f"Email: {details['email']}")
    print(f"Phone: {details['phone']}")
    print(f"Address: {details['address']}")
    print(f"Status: {details['status'].upper()}")
    print(f"Registration Date: {details['registration_date']}")
    print(f"\nRegistered Products:")
    if details['products']:
        for product in details['products']:
            print(f"  • {product['product_name']} (ID: {product['product_id']}) - Premium: ${product['premium']:.2f}")
    else:
        print("  No products registered yet.")
    print(f"Total Monthly Premium: ${details['total_premium']:.2f}")
    print(f"Payments Made: {details['payment_count']}")

def main():
    
    
    display_header("INSURANCE POLICY MANAGEMENT SYSTEM")
    
    print("\n[Step 1: Creating Insurance Products]")
    
    health_product = Product(
        product_name="Premium Health Insurance",
        product_type=ProductType.HEALTH_INSURANCE,
        premium=150.00,
        coverage_amount=100000.00,
        description="Comprehensive health coverage including hospitalization and outpatient care"
    )
    print(f"✓ Created: {health_product}")
    
    car_product = Product(
        product_name="Comprehensive Car Insurance",
        product_type=ProductType.CAR_INSURANCE,
        premium=120.00,
        coverage_amount=250000.00,
        description="Full coverage including theft, accidents, and third-party liability"
    )
    print(f"✓ Created: {car_product}")
    
    life_product = Product(
        product_name="Life Insurance Plus",
        product_type=ProductType.LIFE_INSURANCE,
        premium=200.00,
        coverage_amount=500000.00,
        description="Life insurance with investment benefits"
    )
    print(f"✓ Created: {life_product}")
    

    display_header("CREATING POLICYHOLDERS")
    
    policyholder1 = Policyholder(
        name="John Smith",
        email="john.smith@email.com",
        phone="555-0101",
        address="123 Main St, New York, NY 10001"
    )
    print(f"✓ Created: {policyholder1}")
    
    policyholder2 = Policyholder(
        name="Sarah Johnson",
        email="sarah.johnson@email.com",
        phone="555-0102",
        address="456 Oak Ave, Los Angeles, CA 90001"
    )
    print(f"✓ Created: {policyholder2}")
    
   
    display_header("REGISTERING PRODUCTS FOR POLICYHOLDERS")
    
    print("\n[Registering products for John Smith]")
    policyholder1.register_product(health_product)
    policyholder1.register_product(car_product)
    
    print("\n[Registering products for Sarah Johnson]")
    policyholder2.register_product(health_product)
    policyholder2.register_product(life_product)
    
    
    display_header("PROCESSING PAYMENTS")
    
    print("\n[Processing payments for John Smith]")
    
   
    payment1 = Payment(
        policyholder=policyholder1,
        product=health_product,
        amount=150.00,
        due_date=datetime.now() - timedelta(days=5) 
    )
    print(f"Created: {payment1}")
    payment1.check_overdue()
    payment1.process_payment()
    
    payment2 = Payment(
        policyholder=policyholder1,
        product=car_product,
        amount=120.00,
        due_date=datetime.now() + timedelta(days=10)  
    )
    print(f"Created: {payment2}")
    payment2.send_payment_reminder()
    payment2.process_payment()
    
    print("\n[Processing payments for Sarah Johnson]")
    
   
    payment3 = Payment(
        policyholder=policyholder2,
        product=health_product,
        amount=150.00,
        due_date=datetime.now() + timedelta(days=5)  
    )
    print(f"Created: {payment3}")
    payment3.send_payment_reminder()
    payment3.process_payment()
    
    payment4 = Payment(
        policyholder=policyholder2,
        product=life_product,
        amount=200.00,
        due_date=datetime.now() - timedelta(days=3)  
    )
    print(f"Created: {payment4}")
    payment4.check_overdue()
    payment4.process_payment()
    
    display_header("POLICYHOLDER ACCOUNT DETAILS - JOHN SMITH")
    display_account_details(policyholder1)
    
    display_header("POLICYHOLDER ACCOUNT DETAILS - SARAH JOHNSON")
    display_account_details(policyholder2)
    
    display_header("DEMONSTRATING POLICY MANAGEMENT FEATURES")
    
    print("\n[Testing Suspension and Reactivation]")
    print(f"\nCurrent status of {policyholder1.name}: {policyholder1.status.value}")
    policyholder1.suspend(reason="Non-payment of premium")
    print(f"Status after suspension: {policyholder1.status.value}")
    policyholder1.reactivate()
    print(f"Status after reactivation: {policyholder1.status.value}")
    
    print("\n[Testing Product Updates]")
    print(f"\nCurrent premium for {health_product.product_name}: ${health_product.premium:.2f}")
    health_product.update_premium(175.00)
    print(f"New premium: ${health_product.premium:.2f}")
    
    print(f"\nCurrent coverage for {health_product.product_name}: ${health_product.coverage_amount:.2f}")
    health_product.update_coverage(150000.00)
    print(f"New coverage: ${health_product.coverage_amount:.2f}")
    
    print("\n[Testing Product Suspension]")
    print(f"Current status of {car_product.product_name}: {'Active' if car_product.active else 'Suspended'}")
    car_product.suspend()
    print(f"Status after suspension: {'Active' if car_product.active else 'Suspended'}")
    car_product.reactivate()
    print(f"Status after reactivation: {'Active' if car_product.active else 'Suspended'}")
    

    display_header("PAYMENT TRANSACTION DETAILS")
    
    print("\n[Payment 1 - John Smith's Health Insurance]")
    print(json.dumps(payment1.get_payment_details(), indent=2))
    
    print("\n[Payment 2 - John Smith's Car Insurance]")
    print(json.dumps(payment2.get_payment_details(), indent=2))
    
    print("\n[Payment 3 - Sarah Johnson's Health Insurance]")
    print(json.dumps(payment3.get_payment_details(), indent=2))
    
    print("\n[Payment 4 - Sarah Johnson's Life Insurance]")
    print(json.dumps(payment4.get_payment_details(), indent=2))
    
    # ==================== DISPLAY PRODUCTS ====================
    display_header("INSURANCE PRODUCTS CATALOG")
    
    print("\n[Health Insurance Product]")
    print(json.dumps(health_product.get_product_details(), indent=2))
    
    print("\n[Car Insurance Product]")
    print(json.dumps(car_product.get_product_details(), indent=2))
    
    print("\n[Life Insurance Product]")
    print(json.dumps(life_product.get_product_details(), indent=2))
    
    display_header("SYSTEM DEMONSTRATION COMPLETE")
    print("\n✓ All features demonstrated successfully!")
    print("✓ Policyholders registered with products")
    print("✓ Payments processed and tracked")
    print("✓ Account details and transaction history available")

if __name__ == "__main__":
    main()
```
